```python
class Election:
    election_id = ""
    seat_count = 0
    #mode??
    candidates = []
    ballots = []
    tie_break_order = []
    max_ranks_allowed = 0
    def __init__(self, election_id, seat_count, mode, candidates, ballots, tie_break_order, max_ranks_allowed):
        self.election_id=election_id
        self.seat_count = seat_count
        self.mode = mode
        self.candidates = candidates
        self.ballots = ballots
        self.tie_break_order = tie_break_order
        self.max_ranks_allowed = max_ranks_allowed
```

```python
class Candidate:
    candidate_id = ""
    withdrawn = False
    name = ""
    party = ""
    def __init__(self, candidate_id, withdrawn):
        self.candidate_id = candidate_id
        self.withdrawn = withdrawn
```

```python
class Ballot:
    ballot_id = ""
    rankings = []
    def __init(self, ballot_id, rankings):
        self.ballot_id = ballot_id
        self.rankings = rankings
```

```python
def countBallots(election):
    candidates = {}
    for ballot in election.ballots:
        candidates[ballot.rankings[0].candidate_ids[0]]+=1
    return candidates
        
```
